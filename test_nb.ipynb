# Regression Using StatsModels

```python
import statsmodels.api as sm
import pandas as pd
import numpy as np
import yfinance as yf
```

```python
univ = ['SPY', 'META']
data = yf.download(univ, start='2025-12-01')
```

```python
data.head(5)
```

```python
#! auto adjust is true by default
```

```python
type(data['Close']['META'])
```

```python
adj_close_data = data['Close']
ret = (adj_close_data / adj_close_data.shift()) - 1
```

```python
ret.head()
```

```python
X = ret[['SPY']]
X = sm.add_constant(X)

Y = ret['META']
```

```python
X.head(5)
```

```python
Y.head()
```

```python
X.dropna(inplace=True)
Y.dropna(inplace=True)
```

```python
results = sm.OLS(Y, X).fit()
```

```python
results.summary()
```

```python
results.params
```

```python
META_REC = (results.params['SPY']*X['SPY']) + results.params['const'] + results.resid
```

```python
(Y - META_REC).abs().sum()
```

```python
results.tvalues
```

```python
beta_contr = results.params['SPY'] * X['SPY']
alpha_contr = results.params['const'] + results.resid
```

```python
alpha_contr.cumsum().plot()
```

```python
X['SPY'].corr(alpha_contr) # as expected, Cov(X, alpha + epsilon) = 0
```
